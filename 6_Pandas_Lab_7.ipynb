```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Retail_Dataset_Cleaned.csv")
```

```python
revenue_skew = df["Revenue"].skew()
print(revenue_skew)
```

```python
if revenue_skew > 0:
    print("Positive Skew")
elif revenue_skew < 0:
    print("Negative Skew")
else:
    print("Symmetric Distribution")
```

```python
units_skew = df["Units_Sold"].skew()
print(units_skew)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df["Revenue"], bins=20)
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```
